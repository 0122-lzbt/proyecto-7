```python
import pandas as pd
import plotly.express as px

# Cargar el archivo csv en un DataFrame de pandas
df = pd.read_csv('../vehicles_us.csv')
# Muestra las primeras filas del DataFrame
print(df.head(10))
```

```python
# Crear la columna 'manufacturer' a partir de la columna 'model'
df['manufacturer'] = df['model'].str.split(' ').str[0]
print(df.head())
```

```python
# Crear un histograma de la columnna 'condition' vs 'model_year'
fig = px.histogram(df, x='model_year', color='condition', title='Histograma de condición vs año del modelo')
fig.show()
```

```python
# Crear un histograma de la columna manufacturer vs la columna type, mostrando la cantidad de vehículos por tipo para cada fabricante
fig = px.histogram(df, x='manufacturer', color='type', title='Cantidad de vehículos por tipo para cada fabricante')
fig.show()
```

```python
# Crear un grafico de dispersion de la columna 'price' vs 'odometer', mostrando la relación entre el precio y el kilometraje de los vehículos
fig = px.scatter(df, x='odometer', y='price', color='condition', title='Relación entre precio y kilometraje de los vehículos')
fig.show()
```

```python
# Crear un grafico de dispersión de 'price' vs 'manufacturer', mostrando la relación entre el precio y el fabricante de los vehículos
fig = px.scatter(df, x='manufacturer', y='price', color='condition', title='Relación entre precio y fabricante de los vehículos')
fig.show()
```

```python
# Calculo del precio promedio por fabricante
avg_price_by_manufacturer = df.groupby('manufacturer')['price'].mean().reset_index()
print(avg_price_by_manufacturer)
```
